```python
#수시합불 반별 자료를 하나로 통합: 첫번째 파일만 머리글(header)만 남기고 나머지 엑셀파일의 머리글은 모두 제외함!

import pandas as pd
import os

def merge_excel_files(file_path, file_format, save_path):
    merge_df = pd.DataFrame()
    file_list = [os.path.join(file_path, file) for file in os.listdir(file_path) if file.endswith(file_format)]
    first_file = True

    for file in file_list:
        if file_format == ".xlsx":
            file_df = pd.read_excel(file)
        elif file_format == ".csv":
            file_df = pd.read_csv(file)
        else:
            print(f"Unsupported file format: {file_format}")
            return

        if first_file:
            merge_df = file_df
            first_file = False
        else:
            merge_df = pd.concat([merge_df, file_df.iloc[2:]], ignore_index=True) #세번째 행(0, 1, 2)부터 데이터를 불러옴 - 합불자료의 특성
           #merge_df = pd.concat([merge_df, file_df.iloc[1:]], ignore_index=True) #머릿글이 첫번째 줄에 있을때 이 코드를 사용하면 됨

    if file_format == ".xlsx":
        merge_df.to_excel(save_path, index=False)
    elif file_format == ".csv":
        merge_df.to_csv(save_path, index=False)
    else:
        print(f"Unsupported file format: {file_format}")
        return

    print(f"Merged {len(file_list)} files into {save_path}")

if __name__ == "__main__":
    merge_excel_files(file_path="/content/drive/MyDrive/python/sinmok/data/s", file_format=".xlsx",
                      save_path="/content/drive/MyDrive/python/sinmok/merge_sinmok22_susi.xlsx")
```
